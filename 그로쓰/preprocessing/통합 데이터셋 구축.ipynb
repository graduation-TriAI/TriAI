```python
import numpy as np
import pandas as pd
```

### 데이터 불러와서 지진명 추가

```python
def load_with_name(path, name, n=360):
    f = np.load(path)
    return {k: f[k] for k in f.files} | {"name": np.array([name] * len(f["data"]))}
```

```python
def read_with_name(path, name):
    df = pd.read_csv(path)
    df["name"] = name
    return df
```

### PGV 데이터셋

```python
tohoku_p = read_with_name("/content/drive/MyDrive/TriAI/그로쓰/Data/processed/multimodal/추가 실험/tohoku/new_360_180_pgv.csv", "tohoku")
noto_p = read_with_name("/content/drive/MyDrive/TriAI/그로쓰/Data/processed/multimodal/추가 실험/noto/new_360_180_pgv.csv", "noto")
kumamoto_p = read_with_name("/content/drive/MyDrive/TriAI/그로쓰/Data/processed/multimodal/추가 실험/kumamoto/new_360_180_pgv.csv", "kumamoto")
hokkaido_p = read_with_name("/content/drive/MyDrive/TriAI/그로쓰/Data/processed/multimodal/추가 실험/hokkaido/new_360_180_pgv.csv", "hokkaido")
```

```python
pgv = [tohoku_p, noto_p, kumamoto_p, hokkaido_p]
dataset_names = ["tohoku_p", "noto_p", "kumamoto_p", "hokkaido_p"]

for name, df in zip(dataset_names, pgv):
    print(f"--- Info for {name} ---")
    print(f"Shape: {df.shape}")
    print(f"Columns: {df.columns.tolist()}")
    print(df.dtypes)
    print("-" * 40)
```

### 순서 유지 통합

```python
def merge_datasets(*datasets):
    keys = datasets[0].keys()
    return {k: np.concatenate([d[k] for d in datasets], axis=0) for k in keys}

seismic_all = merge_datasets(tohoku_s, noto_s, kumamoto_s, hokkaido_s)
gnss_all    = merge_datasets(tohoku_g, noto_g, kumamoto_g, hokkaido_g)
pgv_all     = merge_datasets(tohoku_p, noto_p, kumamoto_p, hokkaido_p)

# 순서 검증: name 배열이 세 modality 모두 동일해야 함
assert np.array_equal(seismic_all["name"], gnss_all["name"]), "seismic-gnss 순서 불일치!"
assert np.array_equal(seismic_all["name"], pgv_all["name"]),  "seismic-pgv 순서 불일치!"
assert np.array_equal(gnss_all["name"], pgv_all["name"]),     "gnss-pgv 순서 불일치!"

print("순서 검증 통과")
print("전체 샘플 수:", len(seismic_all["name"]))
print("name 분포:", np.unique(seismic_all["name"], return_counts=True))
```

### 저장

```python
save_dir = "/content/drive/MyDrive/TriAI/그로쓰/Data/processed/multimodal/추가 실험/통합/"

seismic_save = np.savez(save_dir + "seismic_all.npz", **seismic_all)
gnss_save = np.savez(save_dir + "gnss_all.npz",    **gnss_all)
pgv_save = pd.DataFrame(pgv_all).to_csv(save_dir + "pgv_all.csv", index=False)
```
